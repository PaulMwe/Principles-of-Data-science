```python
import pandas as pd
df = pd.read_csv("data/frailty_raw.csv")

## a.	Unit standardization
df["Height_m"] = df["Height_in"] * 0.0254
df["Weight_kg"] = df["Weight_lb"] * 0.45359237
## b.	Feature engineering
df["BMI"] = (df["Weight_kg"] / (df["Height_m"] ** 2)).round(2)
df["AgeGroup"] = pd.cut(df["Age_yr"], [-float("inf"),29,45,60,float("inf")],
                        labels=["<30","30–45","46–60",">60"])
## c.	Categorical → numeric encoding
df["Frailty_binary"] = df["Frailty"].map({"Y":1,"N":0}).astype("int8")
dummies = pd.get_dummies(df["AgeGroup"], prefix="AgeGroup", dtype="int8")
df = pd.concat([df,dummies],axis=1)

##
### To proced with the analyses, i saved the changed in a new csv file
#### frailty_clean.csv

df.to_csv("data/frailty_clean.csv",index=False)
# we will perform some stat evaluation
summary = df.select_dtypes("number").agg(["mean","median","std"]).T
# we will get the correlation
corr = df["Grip_kg"].corr(df["Frailty_binary"])
print(summary)
print("Grip/Frailty correlation:", corr)
```

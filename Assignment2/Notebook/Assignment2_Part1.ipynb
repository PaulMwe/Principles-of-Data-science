```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_DIR = Path("/content")
TRAIN_PATH = DATA_DIR / "train.csv"
df = pd.read_csv(TRAIN_PATH)
df.head()
```

```python
# Count missing values in each column
missing_values = df.isnull().sum()

print(missing_values)
# To see the percentage of missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python

# Convert columns to numerical values
for column in ["Mileage", "Engine", "Power"]:
    df[column] = pd.to_numeric(
        df[column].astype(str).str.split().str[0],
        errors="coerce" )

# Check the converted values
display(df[["Mileage", "Engine", "Power"]].head())
```

```python

#check the missing after the cleaning
display(df[["Mileage", "Engine", "Power"]].isna().sum())
```

```python
# Replace missing values with the median
for column in ["Mileage", "Engine", "Power", "Seats"]:
    df[column] = df[column].fillna(df[column].median())
# Verify missing values again
print("Missing values after cleaning:")
print(df.isnull().sum())

#check the missing after the median
display(df[["Mileage", "Engine", "Power"]].isna().sum())
```

A)**Justification** :

 I examined all columns in the dataset to identify missing values. I decided to use median imputation for the numerical columns Mileage, Engine, Power and Seats because the median is less affected by extreme values and outliers than the mean.

```python
#1. Select – Choose specific columns

# Selected specific columns
selected_data = df[[
    "Name",
    "Location",
    "Year",
    "Owner_Type",
    "Kilometers_Driven",
    "Price"
]]

display(selected_data.head(10))
```

```python
# Filter cars manufactured from 2015 onwards
filtered_data = df[df["Year"] >= 2015]

display(filtered_data.head(10))
```

```python
# Create Vehicle_Age
renamed_data["Vehicle_Age"] = 2026 - renamed_data["Year"]

# Rename columns
renamed_data = df.rename(columns={
    "Owner_Type": "Ownership_Type",
    "Price": "Selling_Price",
    "Location": "Vehicle_Location",
    "Name": "Vehicle_Name"
})

# Display the first 5 rows
display(renamed_data.head(10))
```

```python
# Create a new feature
# This will provide the
renamed_data["Vehicle_Age"] = 2026 - renamed_data["Year"]

display(renamed_data[["Vehicle_Name", "Year", "Vehicle_Age"]].head(10))
```

```python
# Arrange cars by selling price in descending order
arranged_data = renamed_data.sort_values(
    by="Selling_Price",
    ascending=False
)

display(arranged_data.head(10))
```

```python
# Group by Vehicle_Age and calculate average selling price

price_by_age = renamed_data.groupby("Vehicle_Age").agg(
    Total_Vehicles=("Vehicle_Name", "count"),
    Average_Selling_Price=("Selling_Price", "mean")
).reset_index()

# Arrange by vehicle age
price_by_age = price_by_age.sort_values("Vehicle_Age")

# Display the results
display(price_by_age)
```

```python
# Compare selling price for each individual vehicle

price_by_car = renamed_data[[
    "Vehicle_Name",
    "Vehicle_Age",
    "Selling_Price"
]].sort_values(
    by="Vehicle_Age",
    ascending=True
)

# Display the first 20 vehicles
display(price_by_car.head(20))
```

```python
# Group by Vehicle_Age and calculate average selling price
# Grouping the car by age and getting the average selling price
price_by_age = renamed_data.groupby("Vehicle_Age").agg(
    Total_Vehicles=("Vehicle_Name", "count"),
    Average_Selling_Price=("Selling_Price", "mean")
).reset_index()

# Arrange by vehicle age
price_by_age = price_by_age.sort_values("Vehicle_Age")

# Display the results
display(price_by_age)
```

**Interpretation**

The table shows the relationship between vehicle age and average selling price. Each vehicle age is grouped together, and the average selling price is calculated for all vehicles within that age category.

The Total_Vehicles column indicates how many vehicles belong to each age group, while Average_Selling_Price represents their average market price.

This analysis helps us observe how vehicle prices vary as vehicles become older. It also allows us to compare the average selling prices of newer and older vehicles.

**Conclusion**

Based on this analysis, vehicle age can be an important factor influencing selling price. Generally, newer vehicles tend to have higher selling prices than older vehicles because they usually have less depreciation.

However, the selling price does not depend on age alone. Other factors, such as vehicle brand, model, mileage, fuel type, transmission and condition, can also influence its market value.

Overall, grouping vehicles by age provides useful insights into price depreciation and helps us understand how vehicle age relates to resale value.
